# Automated File Sorter

```python
import os , shutil
```

```python
path = r"C:/Users/admin/Desktop/CodeJam/Data Analytics/Python/Automated_File_Sorter_Project/"
```

```python
file_name = os.listdir(path) # List All Directories
```

```python
# os.path.exists(path + 'csv files') # Return False as it ain't exists, now in next shell we will create 3 folder
folder_name = ['csv files','text files','image files']

for i in range(len(folder_name)):
    if not os.path.exists(path + folder_name[i]): # If not exists we will create
        os.makedirs((path + folder_name[i])) # Make New Directories
        #print(path + folder_name[i]) # Print the path name of newly created folders

# file_name has all the directories of the folder acting as a list
for file in file_name: 
    if ".xlsx" in file and not os.path.exists(path + "csv files/" + file): # folder me ho main aur csv files ke andar na ho
        #This will move the xlsx file from main to csv file folder
        shutil.move(path + file , path + "csv files/" + file) # .move(initial path, next path)

    elif ".png" in file and not os.path.exists(path + "image files/" + file): # folder me ho main aur image files ke andar na ho
        #This will move the png file from main to image file folder
        shutil.move(path + file , path + "image files/" + file) # .move(initial path, next path)

    elif ".txt" in file and not os.path.exists(path + "text files/" + file): # folder me ho main aur text files ke andar na ho
        #This will move the txt file from main to text file folder
        shutil.move(path + file , path + "text files/" + file) # .move(initial path, next path)
```
