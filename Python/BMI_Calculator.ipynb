# BMI Calculator

```python
def bmi_calc(w,h):
    # Changed 'and' to 'or' to check if ANY value is invalid, not both
    if(w <= 0) or (h <= 0):
        print("Enter a valid weight and height 🚨")
        # Automatically ask for new input when invalid values are detected
        w = float(input("Re-Enter Patient Weight in Kg:"))
        h = float(input("Re-Enter Patient Height in Mtr:"))
        # Recursively call the function with new values
        return bmi_calc(w, h)
    
    bmi = round(w/(h*h),2)
    print("Your BMI rating is: ",end="")
    if bmi < 16:
        print("Severe Thinness")
    elif 16 < bmi< 17:
        print("Moderate Thinness")
    elif 17 < bmi < 18.5:
        print("Moderate Thinness")
    elif 18.5 < bmi < 25:
        print("Normal")
    elif 25 < bmi < 30:
        print("Overweight")
    elif 30 < bmi < 35:
        print("Obese Class I")
    elif 35 < bmi < 40:
        print("Obese Class II")
    else:
        print("Obese Class III")
    return bmi

if __name__ == "__main__":
    name = str(input("Patient Name:"))
    w = float(input("Patient weight in Kg:"))
    h = float(input("Patient height in mtr:"))
    
    # The function now handles invalid input automatically
    result = bmi_calc(w,h)
    print("Your BMI is:",result,"kg/mtr^2")
    print("HAPPY HEALTHY LIFE ❤")
```
