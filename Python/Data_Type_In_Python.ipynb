# String Type , Exploration & Multiline Comment

```python
# String Type & Multiline Comment
print(type("Hello"),type("2"),type("12.5"),type("False")) # For each it will give (str, str, str, str)

"""
Hello World # This is a multiline comment, assigned withput any variable

"""

multiline_to_str = """ Hi World"""
print(type(multiline_to_str)) # <class 'str'>
```

```python
# String Exploration
a = "Hello World!"
print(a[:5]) # It will print the first 5 letter i.e 0-4 index  - Hello excluding the 5th i.e is a space

print(a[0:9]) # This include the 0 index also - Hello Wor

print(a[6:len(a)]) # 6 to the length of the string - World!

print(len(a)) # It give the length of the string a - 12

print(a[0:5] + a[6:len(a)]) # Concate of string with no space as the 5 is excluded - HelloWorld!

print(a*3) #String multiplcation - Hello World!Hello World!Hello World!

print(a.strip("!"))  # Strip: Remove exclamation mark from beginning/end - "Hello World"
```

# Sets Type & Its Exploration

```python
"""
Python sets are unordered collections of unique elements. They are mutable (you can add or remove items)
, but they do not allow duplicate values and cannot be indexed.
"""
daily_paints = {1,2,3}
type(daily_paints) # set

print(daily_paints) # {1,2,3}

daily_paints_log = {1,2,3,6,8,6,3}
print(daily_paints_log) # Only print the unique values - {1, 2, 3, 6, 8}


daily_paints_log.union(daily_paints) # Only add the unique {1, 2, 3, 6, 8}

daily_paints.issubset(daily_paints_log) # Return ture or false - True

daily_paints.issuperset(daily_paints_log) # False 

daily_paints_log.issuperset(daily_paints) # True

daily_paints.difference(daily_paints_log) # set() empty as smaller one se compare kr rahe hai

daily_paints_log.difference(daily_paints) # It give the difference as it is a super set - {6, 8}

daily_paints.add(4) # It add element in the last 
daily_paints # {1, 2, 3, 4}
```

# Dictionaries Type & Its Exploration

```python
# Key-Pair Values
dict_sec = {"name" : "Vivek Pandey", "age" : 23,"favourite_sub" : ["CS","Maths","CA"]}

type(dict_sec) # dict
dict_sec # {'name': 'Vivek Pandey', 'age': 23, 'favourite_sub': ['CS', 'Maths', 'CA']

dict_sec["name"] # 'Vivek Pandey'

dict_sec["favourite_sub"][0] # 'CS'

dict_sec.items() # dict_items([('name', 'Vivek Pandey'), ('age', 23), ('favourite_sub', ['CS', 'Maths', 'CA'])])

dict_sec.keys() # dict_keys(['name', 'age', 'favourite_sub'])

dict_sec.values() # dict_values(['Vivek Pandey', 23, ['CS', 'Maths', 'CA']])

'''
for key in dict_sec:
    print(key)
    print(dict_sec[key])
#Output:
name
Vivek Pandey
-------
age
23
-------
favourite_sub
['CS', 'Maths', 'CA']
'''
dict_sec["gender"] = "M" # This add new element
dict_sec # {'name': 'Vivek Pandey','age': 23,'favourite_sub': ['CS', 'Maths', 'CA'],'gender': 'M'}

# Update here do not change the value of the exsisting keys, it just add the new key-value pair at the end
dict_sec.update({'name': 'Vivek Pandey','age': 23,'favourite_sub': ['CS', 'Maths', 'CA'],'gender': 'M',"Address":"Gurugram, Haryana","Pin-Code":122001})
dict_sec # {'name': 'Vivek Pandey', 'age': 23, 'favourite_sub': ['CS', 'Maths', 'CA'], 'gender': 'M', 'Address': 'Gurugram, Haryana', 'Pin-Code': 122001}

dict_sec.values() # dict_values(['Vivek Pandey', 23, ['CS', 'Maths', 'CA'], 'M', 'Gurugram, Haryana', 122001])
dict_sec.keys() # dict_keys(['name', 'age', 'favourite_sub', 'gender', 'Address', 'Pin-Code'])

del dict_sec['gender'] # Delete the specific keyfrom the dictionary
dict_sec # {'name': 'Vivek Pandey', 'age': 23, 'favourite_sub': ['CS', 'Maths', 'CA'], 'Address': 'Gurugram, Haryana', 'Pin-Code': 122001}
```
