```python
# Co-relations In Pandas
corr = df.select_dtypes(include=['number']).corr()  # Only select numeric columns for correlation analysis
corr
```

```python
count = df.groupby("Continent")["Rank"].count() # Similar to the sql group by then getting the count continent wise
count.sort_values()
```

```python
mean_1 = df.groupby("Continent").mean(numeric_only=True)
mean_1.sort_values(by = "Rank")
```

```python
df[df["Continent"].str.contains("Oceania")]
```

```python
mean_2 = df.groupby("Continent").mean(numeric_only=True).sort_values(by = "2022 Population", ascending = False)
mean_2
```

```python
mean_trans = mean_2.transpose()
mean_trans
```

```python
mean_trans.plot(figsize = (20,5), marker='o', markersize=8, linestyle='--')
```

```python
mean_2.plot.pie(title = "2022 Popultion Growth Trend",y = "2022 Population", subplots=True,figsize = (8,8))
```

```python
df3 = df.groupby("Continent")[['1980 Population',
       '1990 Population', '2000 Population', '2010 Population',
       '2015 Population', '2020 Population', '2022 Population']].mean(numeric_only= True)

# Another Way = df.groupby("Continent")[df.columns[5:13]].mean(numeric_only= True)

# As we only want the populations of different years
df3
```

```python
new_df = df3.transpose()
new_df
```

```python
new_df.plot(title = "Population Growth Trend", figsize = (20,5), marker='o', markersize=8, linestyle='--')
# We cann see the growth from 1980 to 2022, lasrly asia and africa raised the population
```

```python
sb.heatmap(new_df, annot = True, linewidths=3,  linecolor='black')
plt.rcParams['font.serif'] = 'Times New Roman'
plt.show()
# from top to bottom the asia -> pink(low)-1980 to white(high)-2022
```
