# Find_all & Find

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://www.scrapethissite.com/pages/forms/"
```

```python
page = requests.get(url)
```

```python
soup = BeautifulSoup(page.text,"html")
```

```python
soup.find('div') # only give the required tag that appeared first
```

```python
soup.find_all('div',class_ = "col-md-12")
```

```python
soup.find_all('p', class_ = "lead") # This give in term of tag and the text inside it
```

```python
# But if we just want the text inside that <p> tag we have to use the "find"
print(soup.find('p', class_ = 'lead').text.strip()) # strip() removes the unnecessary white spaces
```

```python
soup.find_all('th')
```

```python
# But we want the only team neame that is the first occurence of the "th" tag
print(soup.find('th').text.strip()) # Team Name
```
