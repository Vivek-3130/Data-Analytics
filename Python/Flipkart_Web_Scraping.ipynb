```python
from bs4 import BeautifulSoup
from curl_cffi import requests
import time
import datetime
import smtplib
```

```python
soup1 = BeautifulSoup(response.content, "html.parser")
#print(soup1)
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")
print(soup2)
```
