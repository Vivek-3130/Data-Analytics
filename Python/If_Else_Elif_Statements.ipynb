# If-Else-Elif Statements & Its Exploration

```python
# If 
if 25 > 10:
    print("Yes It's True")
```

```python
# If-else
if 25 > 50:
    print("If Worked")
else:
    print("Else Worked") # Else Worked
```

```python
# If - Elif - ELse:
if 25 > 50:
    print("If Worked")
elif 25 < 45:
    print("Elif Worked")
else:
    print("Else Worked")
```

```python
# Another way of writing a if statement
print("It Worked") if (10 > 30) else print("It Doesn't Worked") # It Doesn't Worked
```

```python
# Nested If Statement
if (25 < 15) or (45 > 28):
    print("Logical Operator Worked")
    if (25 < 35):
        print("Nested If Worked")
    else:
        print("Nested If Is Wrong")
else:
    print("Logical Operator Ain't Worked")

# Logical Operator Worked
# Nested If Worked - If not worked then given the nested if is wrong as output
```
