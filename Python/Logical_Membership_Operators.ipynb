# Logical Operators Type & Its Exploration

# Membership Operator

```python
10 == 10 # This is an equal operator - True
10 == 9 # False

10 != 10 # Not Equal Operator - False
10 != 9 # True
```

```python
10 > 7 # Greater Than Operator - True
10 < 7 # Less Than Operator - False
```

```python
10 >= 8 # Greater than & equal to - True 
10 <= 10 # Less than & equal to - True

10 >= 11 # False
```

# Logical Operator

```python
# Here we will be seeing and , not , or operator
10 | 5 # This Bitwise OR - Addition - 15 (1111)

10 & 5 # This is Bitwise AND - Multiplication - 0 (0000)

10 ^ 5 # This is Bitwise XOR - Return 1 if 2 bits are opposite - 15(1111)
```

```python
# AND
(10 > 60) and (60 > 50) # Return False as one of the values are false

(80 > 60) and (60 > 50) # Return True as both are true

('Vanilla' > 'Chocklate') and (60 > 50) # Here both are true as V > C return True

# OR
(80 > 60) or (60 > 50) # Here as both are true so returns True

(10 > 60) or (60 > 50) # Here also returns True as one of them is true

# NOT
not((80 > 60) or (60 > 50)) # Here as both are true so returns the opposite as False

not((10 > 60) and (60 > 50)) # Here as one is true and other is false so returns the opposite as True
```

```python
str1 = "I Love India"

"L" in str1 # Return true as L is present in the str1
"P" in str1 # Return false as P is not present in the str1
"i" not in str1 # Return false as I is present in the str1
```
