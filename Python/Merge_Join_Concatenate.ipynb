# Merging , Join And Concatenation

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\admin\Desktop\CodeJam\Data Analytics\Python\Pandas\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\admin\Desktop\CodeJam\Data Analytics\Python\Pandas\LOTR 2.csv")
df2
```

```python
df1.merge(df2) # By Default It Will Be A Inner Join
```

```python
 df1.merge(df2, left_on='FellowshipID', right_on='FellowshipID') # Basically in sql ON e.empid = d.empid common key
```

```python
# Second Option
df1.merge(df2, how = "inner", on = "FellowshipID") # This will give same result as of the previuos one
```

```python
df1.merge(df2, how = "right", on = "FellowshipID")
```

```python
df1.merge(df2, how = "left", on = "FellowshipID")
```

```python
df1.merge(df2, how = "outer", on = "FellowshipID")
```

```python
#Join
df1.join(df2,on = "FellowshipID",  how = "outer", lsuffix="_Left", rsuffix="_Right")
```

```python
df3 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), lsuffix="_Left", rsuffix="_Right", how = "outer")
df3 # We first created the index then join to make it clear
```

```python
# Concatenation
pd.concat([df1,df2],join = "inner",axis = 1) # Axis 1 pr sab dikahyega and zero pr bs index o,1,2,3 column vala
```
