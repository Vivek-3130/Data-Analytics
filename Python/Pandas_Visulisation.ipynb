# Pandas Visualistaion

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
print(plt.style.available) # It give all the tsyle it have an we can use them in our visualisation
plt.style.use('seaborn-v0_8-bright')
```

```python
df = pd.read_csv(r"C:\Users\admin\Desktop\CodeJam\Data Analytics\Python\Pandas\Ice Cream Ratings.csv")
df = df.set_index("Date")
df
```

```python
# Visualistation
df.plot(title = "Line Chart Basic", xlabel = "Monthly Timeline", ylabel = "Ratings")   # It took the date as the xlable on the x- axis
```

```python
df.plot(kind = "bar" , title = "Flavoured Bar Chart", subplots = True, xlabel = "Monthly Timeline", ylabel = "Ratings") # subplots give three diff chart in comparision
```

```python
# Sngle Legend
df["Overall Rating"].plot(kind = 'bar', color = 'red', xlabel = "Monthly Timeline", ylabel = "Ratings")
```

```python
df.plot.barh(color = ["red","blue","black"],xlabel = "Monthly Timeline", ylabel = "Ratings", stacked = True, figsize = (10,5) ) # direct way to visualise
```

```python
# Scattering
df.plot.scatter(x = "Flavor Rating", y = "Overall Rating", s = 55, c='DarkGreen')
```

```python
# Box Plot
df.boxplot() # It reflects the bottom ,mid and the highest value
```

```python
df.plot.area(color = ["yellow","crimson","olive"], stacked = True, figsize = (10,5))
```

```python
df.plot.pie(y = "Overall Rating", figsize = (10,5.5)) #It is necessary to give the y column any
```
