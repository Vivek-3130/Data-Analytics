# Web Scraping Using Python

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue"
```

```python
page = requests.get(url, headers= { 'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/116.0.0.0 Safari/537.36' } )

soup = BeautifulSoup(page.text,"html")

# print(soup.prettify())
```

```python
table = soup.find_all('table')[0] #As html was returning the list so we captured the table through indexing , 1st table of the page will be returned
```

```python
world_titles = table.find_all('th') # This give the list of all the "th" tag element inside the table[1] - yha list aayi
```

```python
world_titles_table = [title.text.strip().upper() for title in world_titles] # Here we are using the list comprehension, and iteratiing inside the list
```

```python
world_titles_table # ['Rank', 'Name', 'Industry', 'Revenue (USD billions)', 'Employees', 'Headquarters'], 1st table col name will be returned
```

# Web Scraping Using Pandas

```python
import pandas as pd
```

```python
df = pd.DataFrame( columns = world_titles_table) # Here we created our data frame

df # Called the data frame :- 	RANK	NAME	INDUSTRY	REVENUE (USD BILLIONS)	EMPLOYEES	HEADQUARTERS
```

```python
# Fetching the data from the table to our dataframe
column_data = table.find_all('tr') # Yha list aayi
```

```python
for rows in column_data[1:]: # 1 ke baad ke sare as 0th to blank list hai
    rows_data = rows.find_all('td') # Yha bhi list generate hui
    individual_row_data = [row.text.strip().upper() for row in rows_data] # List Comprehension

    length = len(df) # created a var to store different rows
    df.loc[length] = individual_row_data # Values going to the current length
```

```python
df
```

```python
df = df.drop_duplicates().reset_index(drop=True) # This removes the duplicate values from the table
```
