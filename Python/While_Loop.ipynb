# While Loop Statements

```python
num = 1
while (num <= 5):
    print(num)
    num += 1 #1,2,3,4,5
```

```python
# While with break
num = 1
while (num <= 5):
    print(num)
    if num == 3:
        break
    num += 1
    else: 
       print("No Longer Applicable") #1,2,3, No Longer Applicable
    
```

```python
# While with continue
num = 1
while (num <= 5):
    num += 1
    if num == 3:
        continue # for continue we have to update the value earlier other wise it will be never ending 3,3,3,3,3.......
        print(num)
    else: 
       print("No Longer Applicable") #1,2,4,5 No Longer Applicable
```
